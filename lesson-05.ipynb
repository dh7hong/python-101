#### String Methods

```python
# Uppercase and Lowercase
name = "Alice"
print(f'Upper case: {name.upper()}')  # Output: ALICE
print(f'Lower case: {name.lower()}')  # Output: alice

# Capitalize
name_2 = "bob"
print(f'Capitalized: {name_2.capitalize()}')  # Output: Bob

# Title Case
full_name = "charlie brown"
print(f'Title Case: {full_name.title()}')  # Output: Charlie Brown

```

#### Some more string methods

```python
# Removing Whitespace
greeting = "   Hello, World!   "
print(f'Original: "{greeting}"')
print(f'Stripped: "{greeting.strip()}"')  # Output: "Hello, World!"

# left strip
print(f'Left Stripped: "{greeting.lstrip()}"')  # Output: "Hello, World!   "
# right strip
print(f'Right Stripped: "{greeting.rstrip()}"')  # Output: "   Hello, World!"

# Replacing Substrings
text = "I like cats. Cats are great!"
print(f'Original: "{text}"')
print(f'Replaced: "{text.replace("cats", "dogs")}"')  # Output: "I like dogs. Cats are great!"

# Finding Substrings
sentence = "The quick brown fox jumps over the lazy dog."
print(f'Index of "fox": {sentence.find("fox")}')  # Output: 16
# If the substring is not found, find() returns -1
print(f'Index of "cat": {sentence.find("cat")}')  # Output: -1
```

#### More String Related

```python
# Checking if Text Exists
sentence = "The quick brown fox jumps over the lazy dog."
print(f'Contains "quick": {"quick" in sentence}')  # Output: True
print(f'Contains "cat": {"cat" in sentence}')  # Output: False

# Splitting and Joining Strings
text = "apple,banana,cherry"
fruits = text.split(",")  # Split the string into a list
print(f'Fruits: {fruits}')  # Output: ['apple', 'banana', 'cherry']

text_for_joining = ["Hello", "World"]
joined_text = " ".join(text_for_joining)  # Join the list into a string
print(f'Joined Text: "{joined_text}"')  # Output: "Hello World"
```

###### Formatting numbers

```python
price = 12.34567

print(f"{price:.2f}")  # Output: 12.35 (rounded to 2 decimal places)
```

#### Checking String Properties

```python
# Is Alphabetic?
print(f'"Python".isalpha(): {"Python".isalpha()}')  # Output: True

# Is Numeric?
print(f'"12345".isdigit(): {"12345".isdigit()}')  # Output: True

# Is Alphanumeric?
print(f'"Python123".isalnum(): {"Python123".isalnum()}')  # Output: True
print(f'"Python!".isalnum(): {"Python!".isalnum()}')  # Output: False

# Counting Occurrences
text = "banana"
print(f'Count of "a": {text.count("a")}')  # Output: 3

# Startswith and Endswith
filename = "report.pdf"
print(f'Filename starts with "rep": {filename.startswith("rep")}')  # Output: True
print(f'Filename ends with ".pdf": {filename.endswith(".pdf")}')  # Output: True

# Reversing a String
text = "Hello"
reversed_text = text[::-1]  # Reverse the string using slicing
print(f'Reversed: "{reversed_text}"')  # Output: "olleH"
```

```python
my_string = "Helloz"
# sort reverse
sorted_string = ''.join(sorted(my_string, reverse=True))
print(f'Sorted in reverse: "{sorted_string}"')  # Output: "roollledWH,! "

# sort vs sorted
my_string = "Helloz"
print(f'Sorted in reverse using sorted(): "{sorted(my_string, reverse=True)}"')  # Output: ['z', 'o', 'l', 'l', 'e', 'H']
my_list = list(my_string)
my_list.sort(reverse=True)
print(f'Sorted in reverse using sort(): "{my_list}"')  # Output: ['z', 'o', 'l', 'l', 'e', 'H']
```

```python
s = "Zbcdefg"

def solution(s):
    return ''.join(sorted(s, reverse=True))

print(solution(s))  # Output: "gfedcbZ"
```
