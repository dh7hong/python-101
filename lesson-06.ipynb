#### More on Functions

#### 1. Local Variables

```python
def example():
    x = 10
    print(x)

example()

# print(x)  # This will raise a NameError because x is not defined in this scope
```

#### 2. Scope

```python
name = "Alice"

def show():
    print(name)

show()
```

#### 3. Modifying Global Variables

```python
count = 0

def update():
    global count
    count += 1

update()

print(count)

update()

print(count)

update()

print(count)
```

#### 4. Functions calling another function

```python
def greet():
    print("Hello")

def welcome():
    greet()
    print("Welcome!")

welcome()
```

#### 5. Lambda Functions - Small one-line functions.

```python
# Normal:
def square(x):
    return x * x

# Lambda:
square_lambda = lambda x: x * x

print(square(5))  # Output: 25
print(square_lambda(5))  # Output: 25
```

#### 6. Docstrings

```python
def add(a, b):
    """
    Returns the sum of two numbers.
    """
    return a + b

help(add)
```
